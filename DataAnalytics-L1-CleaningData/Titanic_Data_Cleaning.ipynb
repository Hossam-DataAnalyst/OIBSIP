```python
# 1. استدعاء المكتبة والداتا
import pandas as pd
url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
df = pd.read_csv(url)

# 2. مسح عمود الكابينة (الطريقة الأحدث)
df = df.drop('Cabin', axis=1)

# 3. ملء فراغات العمر (الطريقة الأحدث)
متوسط_العمر = df['Age'].mean()
df['Age'] = df['Age'].fillna(متوسط_العمر)

# 4. ملء فراغات ميناء الركوب (الطريقة الأحدث)
اكتر_ميناء = df['Embarked'].mode()[0]
df['Embarked'] = df['Embarked'].fillna(اكتر_ميناء)

# 5. التقرير الطبي
df.info()
```

```python
# حفظ الداتا النظيفة في ملف CSV
df.to_csv('cleaned_titanic_data.csv', index=False)

# كود صغير عشان ينزلك الملف أوتوماتيك على الجهاز (لأنك شغال على المتصفح)
from google.colab import files
files.download('cleaned_titanic_data.csv')
```
